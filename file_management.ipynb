```python
import json
import os

def split_json_files(input_folder, output_folder, chunk_size=80):
    # Create the output folder if it doesn't exist
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
        print(f"Created output folder: {output_folder}")

    # Verify input folder exists
    if not os.path.exists(input_folder):
        print(f"Error: Input folder '{input_folder}' not found.")
        return

    # Process each file in the input directory
    for filename in os.listdir(input_folder):
        if filename.endswith(".json"):
            file_path = os.path.join(input_folder, filename)
            base_name = os.path.splitext(filename)[0]

            print(f"Reading {filename}...")

            try:
                with open(file_path, 'r', encoding='utf-8') as f:
                    data = json.load(f)

                # Ensure the root is a list/array
                if not isinstance(data, list):
                    print(f"Skipping {filename}: Data is not an array.")
                    continue

                # Split the data into chunks
                for i in range(0, len(data), chunk_size):
                    part_num = (i // chunk_size) + 1
                    chunk = data[i : i + chunk_size]
                    
                    # Generate name: <original_file_name>_p1.json
                    new_filename = f"{base_name}_p{part_num}.json"
                    new_file_path = os.path.join(output_folder, new_filename)

                    # Save the chunk
                    with open(new_file_path, 'w', encoding='utf-8') as f_out:
                        # indent=4 for readability; ensure_ascii=False to keep CO2, etc., as is
                        json.dump(chunk, f_out, indent=2, ensure_ascii=False)
                    
                print(f"  Successfully split {filename} into {part_num} parts.")

            except Exception as e:
                print(f"  Error processing {filename}: {e}")

if __name__ == "__main__":
    # Settings
    INPUT_DIR = "2016"
    OUTPUT_DIR = "aiche_file_2016_output"
    MAX_ELEMENTS = 80

    split_json_files(INPUT_DIR, OUTPUT_DIR, MAX_ELEMENTS)
    print("\nAll tasks completed.")
```
