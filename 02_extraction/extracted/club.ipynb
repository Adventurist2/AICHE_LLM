```python
import json
import glob

# --- STEP 1: Get all JSON files in the folder ---
# Modify path if needed (example: "./data/*.json")
files = glob.glob("*.json")

big_array = []

for f in files:
    print("Reading:", f)
    with open(f, "r", encoding="utf-8") as infile:
        data = json.load(infile)

        # Each file MUST contain a JSON array
        if not isinstance(data, list):
            raise ValueError(f"File {f} does not contain a JSON array!")

        # Append elements *as they are*
        big_array.extend(data)

# --- STEP 2: Write output file ---
with open("combined.json", "w", encoding="utf-8") as outfile:
    json.dump(big_array, outfile, indent=2, ensure_ascii=False)

print("Done! Combined JSON written to combined.json")
```
